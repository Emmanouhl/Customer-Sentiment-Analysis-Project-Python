```python
import csv
# ==========================================================
# Function: Open file and return reader
# ==========================================================
def open_file(file):
    """ Opens a CSV file and returns a DictReader object """
    csv_file = open(file, newline = "", encoding = "utf-8")
    reader = csv.DictReader(csv_file)
    return reader

# ======================================================
# Task: COUNT ALL COUNTRIES
# ======================================================
def count_country(filename):
    
# Initialize counters for each country
    nigeria_count = 0
    kenya_count = 0
    south_africa_count = 0
    ghana_count = 0
    other_countries_count = 0 #For any other countries not listed

# Get the reader from the function
    reader = open_file(filename)
#Loop through each row and count countries
    for row in reader:
        # Match the "Country" column against the known countries;
        # anything not in the list falls into "other_countries_count".
        if row["Country"] == "Nigeria":
            nigeria_count += 1
        elif row["Country"] == "Kenya":
            kenya_count += 1
        elif row["Country"] == "South Africa":
            south_africa_count += 1
        elif row["Country"] == "Ghana":
            ghana_count += 1
        else:
            other_countries_count += 1
        
# Print the results                       
    print(f"Nigeria feedback records: {nigeria_count}")
    print(f"Kenya feedback records: {kenya_count}")
    print(f"South Africa feedback records: {south_africa_count}")
    print(f"Ghana feedback records: {ghana_count}")
    print(f"Other Countries feedback records: {other_countries_count}")

count_country("customer_feedback.csv")
```

```python
import csv
# ==========================================================
# Function: Open file and return reader
# ==========================================================
def open_file(file):
    """ Opens a CSV file and returns a DictReader object """
    csv_file = open(file, newline = "", encoding = "utf-8")
    reader = csv.DictReader(csv_file)
    return reader

def analyse_channel(filename):
    """ Analyse feedback channels """
    # Reference list of the channels we expect to see (not used for
    # matching directly below, but documents the valid categories).
    channels = [
        "website",
        "app",
        "whatsapp",
        "social media"
    ]
    website_count = 0
    app_count =  0
    whatsapp_count = 0
    social_media_count = 0
    others_count = 0

    total_channel_count = 0

    reader = open_file(filename)
    for row in reader:
        # Normalise case/whitespace so "Website", " website ", etc. all match.
        feedback_channels = row["Channel"].lower().strip()
        if feedback_channels == "website":
            website_count += 1
        elif feedback_channels == "app":
            app_count += 1
        elif feedback_channels == "whatsapp":
            whatsapp_count += 1
        elif feedback_channels == "social media":
            social_media_count += 1
        else:
            others_count += 1
        total_channel_count += 1
        
    # Print each channel's count alongside its share of the total feedback.
    print(f"Website Feedback Records: {website_count} ({(website_count / total_channel_count) * 100:.2f}%)")
    print(f"App Feedback Records: {app_count} ({(app_count / total_channel_count) * 100:.2f}%)")
    print(f"Whatsapp Feedback Records: {whatsapp_count} ({(whatsapp_count / total_channel_count) * 100:.2f}%)")
    print(f"Social Media Feedback Records: {social_media_count} ({(social_media_count / total_channel_count) * 100:.2f}%)")
    print(f"Other Channels Feedback Records: {others_count} ({(others_count / total_channel_count) * 100:.2f}%)")
       

analyse_channel("customer_feedback.csv")

```

```python
import csv
# ==========================================================
# Function: Open file and return reader
# ==========================================================
def open_file(file):
    """ Opens a CSV file and returns a DictReader object """
    csv_file = open(file, newline = "", encoding = "utf-8")
    reader = csv.DictReader(csv_file)
    return reader

# Categorises each feedback record into an issue type based on keywords
# found in the free-text "Feedback_Text" column.
def categorize_issue(filename):
   
    delivery_count = 0
    payment_count = 0
    product_quality_count = 0
    customer_service_count = 0
    refund_count = 0
    technical_issue_count = 0
    other_count = 0

    total_count = 0

    reader = open_file(filename)
    for row in reader:
        feedback_text = row["Feedback_Text"].lower().strip()
        text = feedback_text
         
        # NOTE: the checks below are evaluated in order (elif chain), so a
        # feedback text matching multiple categories will only be counted
        # under the FIRST matching category, not all of them.
        if "delivery" in text or "late" in text or "delay" in text or "arrived" in text:
            issue = "Delivery"
            delivery_count += 1
        elif "payment" in text or "failed" in text or "pay" in text or "transaction" in text:
            issue = "Payment"
            payment_count += 1
        elif "product" in text or "quality" in text or "item" in text or "damage" in text or "broken" in text: 
            issue = "Product Quality"
            product_quality_count += 1
        elif "support" in text or "service" in text or "help" in text or "customer" in text: 
            issue = "Customer Service"
            customer_service_count += 1
        elif "refund" in text or "return" in text or "exchange" in text or "reimbursement" in text:
            issue = "Refund"
            refund_count += 1
        elif "technical" in text or "error" in text or "bug" in text or "issue" in text or "website" in text or "app" in text:
            issue = "Technical Issue"
            technical_issue_count += 1

        else:
            # No keyword matched any category.
            issue = "Others"
            other_count += 1
        total_count += 1
            
            
        
       
    # Print the count of feedback records per issue category.
    print(f"Delivery Issue: {delivery_count}")
    print(f"Payment Issue: {payment_count}")
    print(f"Product Quality Issue: {product_quality_count}")
    print(f"Customer Service Issue: {customer_service_count}")
    print(f"Refund Issue: {refund_count}")
    print(f"Technical Issue: {technical_issue_count}")
    print(f"Other Issue: {other_count}")
    print("------")
    print(f"Total Complaints Issue: {total_count}")

    
categorize_issue("customer_feedback.csv")

    
```

```python
import csv
# ==========================================================
# Function: Open file and return reader
# ==========================================================
def open_file(file):
    """ Opens a CSV file and returns a DictReader object """
    csv_file = open(file, newline = "", encoding = "utf-8")
    reader = csv.DictReader(csv_file)
    return reader
# ==========================================================
# TASK: DELIVERY ANALYSIS
# ==========================================================

def analyse_delivery(filename):
    on_time_count = 0
    slight_delay_count = 0
    moderate_delay_count = 0
    severe_delay_count = 0

    total_deliveries_count = 0

    reader = open_file("customer_feedback.csv")
    for row in reader:
        actual_delivery = int(row["Delivery_Days"])
        expected_delivery = int(row["Expected_Delivery_Days"])
        # Positive delay = number of days late; zero or negative = on time.
        delay = actual_delivery - expected_delivery
        if delay <= 0:
            delivery_status = "On Time"
            on_time_count += 1
        elif delay <= 2:
            delivery_status = "Slight Delay"
            slight_delay_count += 1
        elif delay <= 5:
            delivery_status = "Moderate Delay"
            moderate_delay_count += 1
        else:
            delivery_status = "Severe Delay"
            severe_delay_count += 1
        total_deliveries_count += 1


    # Print each delivery bucket's count and percentage of all deliveries.
    print(f"On Time Deliveries: {on_time_count} ({(on_time_count / total_deliveries_count) * 100:.2f}%)")
    print(f"Slight Delay Deliveries: {slight_delay_count} ({(slight_delay_count / total_deliveries_count) * 100:.2f}%)")
    print(f"Moderate Delay Deliveries: {moderate_delay_count} ({(moderate_delay_count / total_deliveries_count) * 100:.2f}%)")
    print(f"Severe Delay Deliveries: {severe_delay_count} ({(severe_delay_count / total_deliveries_count) * 100:.2f}%)")
    print("--------")
    print(f"Total Deliveries: {total_deliveries_count}")       

    

analyse_delivery("customer_feedback.csv")


    
```

```python
import csv
# ==========================================================
# Function: Open file and return reader
# ==========================================================
def open_file(file):
    """ Opens a CSV file and returns a DictReader object """
    csv_file = open(file, newline = "", encoding = "utf-8")
    reader = csv.DictReader(csv_file)
    return reader
# ==========================================================
# SENTIMENT ANALYSIS SETUP
# ==========================================================
def analyse_sentiment(filename):
    """Analyse customer feedback sentiment and print resolution statistics."""

    # Simple keyword-based sentiment lexicon: any of these words appearing
    # in the feedback text counts as a "vote" for that sentiment.
    positive_words = [
        "excellent",
        "great",
        "good",
        "fast",
        "helpful",
        "love",
        "loved",
        "quick"
    ]

    negative_words = [
        "bad",
        "poor",
        "late",
        "slow",
        "failed",
        "disappointing",
        "disappointed",
        "terrible"
    ]
    positive_total = 0
    negative_total = 0
    neutral_total = 0

    overall_total = 0
 
    reader = open_file(filename)

    for row in reader:
        feedback_text = row["Feedback_Text"].lower().strip()

        # Count how many positive keywords appear in this feedback.
        positive_count = 0
        for word in positive_words:
            if word in feedback_text:
                positive_count += 1

        # Count how many negative keywords appear in this feedback.
        negative_count = 0
        for word in negative_words:
            if word in feedback_text:
                negative_count += 1

        # Determine Sentiment
        # Whichever count is higher wins; a tie (including 0 vs 0) is Neutral.
        if positive_count > negative_count:
            sentiment = "Positive"
            positive_total += 1

        elif negative_count > positive_count:
            sentiment = "Negative"
            negative_total += 1

        else:
            sentiment = "Neutral"
            neutral_total += 1
        overall_total += 1
    # Print sentiment breakdown with percentages of overall feedback.
    print(f"Positive Feedbacks: {positive_total} ({(positive_total / overall_total) * 100:.2f}%)")
    print(f"Negative Feedbacks: {negative_total} ({(negative_total / overall_total) * 100:.2f}%)")
    print(f"Neutral Feedbacks: {neutral_total} ({(neutral_total / overall_total) * 100:.2f}%)")
    print("---------")
    print(f"Overall Total Feedbacks: {overall_total}")  

    
analyse_sentiment("customer_feedback.csv")
    
```

```python
import csv

# ============================================================
# Function: Open file and return reader
# ============================================================
def open_file(file):
    """Opens a CSV file and returns a DictReader object"""
    csv_file = open(file, newline="", encoding="utf-8")
    reader = csv.DictReader(csv_file)
    return reader


# ============================================================
# SENTIMENT ANALYSIS SETUP
# ============================================================
positive_words = [
    "excellent", "great", "good", "fast",
    "helpful", "love", "loved", "quick"
]

negative_words = [
    "bad", "poor", "late", "slow",
    "failed", "disappointing", "disappointed", "terrible"
]

# Issue category keywords
# Maps each issue category to the keywords that identify it in the
# feedback text (used below to classify each complaint).
issue_patterns = {
    "Delivery": ["delivery", "late", "delay", "arrived", "ship"],
    "Payment": ["payment", "pay", "failed", "transaction", "card"],
    "Customer Service": ["support", "service", "help", "staff", "agent"],
    "Product": ["product", "quality", "item", "damaged", "broken"],
    "Refund": ["refund", "return", "reimbursement"],
    "Technical Issue": ["app", "website", "error", "bug", "technical"]
}


# ============================================================
# MAIN ANALYSIS FUNCTION
# Builds on the combined analysis above by also classifying each
# complaint into an issue category and computing a few "key findings"
# (relationships between delay/resolution and negative sentiment).
# ============================================================
def analyse_feedback(filename):

    # ---------- COUNTERS ----------
    positive_total = 0
    negative_total = 0
    neutral_total = 0

    rating_sum = 0
    rating_count = 0

    on_time_count = 0
    slight_delay_count = 0
    moderate_delay_count = 0
    severe_delay_count = 0

    resolved_count = 0
    unresolved_count = 0

    # Issue counters
    issue_counts = {
        "Delivery": 0,
        "Payment": 0,
        "Customer Service": 0,
        "Product": 0,
        "Refund": 0,
        "Technical Issue": 0,
        "Other": 0
    }

    # For key findings
    # Track how often a severe delivery delay / an unresolved complaint
    # coincides with negative sentiment, used in the "KEY FINDINGS" section.
    severe_delay_negative = 0
    severe_delay_total = 0
    unresolved_negative = 0

    # ---------- READ AND PROCESS ----------
    reader = open_file(filename)

    for row in reader:

        # --- Sentiment Analysis ---
        feedback_text = row["Feedback_Text"].lower().strip()

        positive_count = 0
        for word in positive_words:
            if word in feedback_text:
                positive_count += 1

        negative_count = 0
        for word in negative_words:
            if word in feedback_text:
                negative_count += 1

        if positive_count > negative_count:
            sentiment = "Positive"
            positive_total += 1
        elif negative_count > positive_count:
            sentiment = "Negative"
            negative_total += 1
        else:
            sentiment = "Neutral"
            neutral_total += 1

        # --- Rating ---
        rating = int(row["Rating"])
        rating_sum += rating
        rating_count += 1

        # --- Delivery Analysis ---
        actual_delivery = int(row["Delivery_Days"])
        expected_delivery = int(row["Expected_Delivery_Days"])
        delay = actual_delivery - expected_delivery

        if delay <= 0:
            on_time_count += 1
        elif delay <= 2:
            slight_delay_count += 1
        elif delay <= 5:
            moderate_delay_count += 1
        else:
            severe_delay_count += 1
            severe_delay_total += 1
            if sentiment == "Negative":
                severe_delay_negative += 1

        # --- Resolution Analysis ---
        if row["Resolution_Status"] == "Resolved":
            resolved_count += 1
        else:
            unresolved_count += 1
            if sentiment == "Negative":
                unresolved_negative += 1

        # --- Issue Categorization ---
        # Walk through each issue category's keyword list and stop at the
        # first category that matches (mirrors the standalone issue-category
        # cell above); anything unmatched is counted as "Other".
        matched = False
        for issue, keywords in issue_patterns.items():
            for keyword in keywords:
                if keyword in feedback_text:
                    issue_counts[issue] += 1
                    matched = True
                    break
            if matched:
                break
        if not matched:
            issue_counts["Other"] += 1

    # ---------- CALCULATIONS ----------
    total_feedback = positive_total + negative_total + neutral_total
    average_rating = rating_sum / rating_count if rating_count > 0 else 0
    negative_rate = (negative_total / total_feedback) * 100 if total_feedback > 0 else 0

    # Determine top issues (excluding "Other"), most frequent first.
    top_issues = sorted(
        [(k, v) for k, v in issue_counts.items() if k != "Other"],
        key=lambda x: x[1],
        reverse=True
    )

    # ---------- PRINT REPORT ----------
    print("==================================================")
    print("        AFRICART CUSTOMER SENTIMENT ANALYSIS")
    print("==================================================")
    print()

    print("DATASET SUMMARY")
    print("..................................................")
    print(f"Total Feedback: {total_feedback}")
    print()

    print("SENTIMENT")
    print("..................................................")
    print(f"Positive: {positive_total}")
    print(f"Neutral: {neutral_total}")
    print(f"Negative: {negative_total}")
    print()
    print(f"Negative Rate: {negative_rate:.2f}%")
    print()

    print("CUSTOMER RATING")
    print("..................................................")
    print(f"Average Rating: {average_rating:.2f}")
    print()

    print("DELIVERY")
    print("..................................................")
    print(f"On Time: {on_time_count}")
    print(f"Slight Delay: {slight_delay_count}")
    print(f"Moderate Delay: {moderate_delay_count}")
    print(f"Severe Delay: {severe_delay_count}")
    print()

    print("RESOLUTION")
    print("..................................................")
    print(f"Resolved: {resolved_count}")
    print(f"Unresolved: {unresolved_count}")
    print()

    print("TOP CUSTOMER ISSUES")
    print("..................................................")
    for issue, count in top_issues[:4]:
        print(issue)
    print()

    print("KEY FINDINGS")
    print("..................................................")

    # The findings below were derived by manually inspecting the dataset
    # (Feedback_IDs, counts and percentages are hard-coded, not computed
    # live from the counters above) to summarise the most notable patterns.
    print(f"""
    1. Overall customer sentiment is positive but fragile:
   - 50% Positive (15 of 30)
   - 10% Neutral (3 of 30)
   - 40% Negative (12 of 30)
   
   Nearly half of all feedback is negative — this is a warning sign.
   
   2. Negative sentiment is highest among severe delays.
   - Severe Delay: 1 record with 100% negative sentiment
   - The one severe delay case (FB014, 10 days vs 4 expected)
     resulted in "Very bad experience and late delivery"
   
   → Delivery delay is a direct driver of customer dissatisfaction.

   3. Unresolved complaints contain a higher proportion of negative feedback.
   - Unresolved: 11 complaints, 100% Negative (11 of 11)
   - Resolved: 19 complaints, only 5.3% Negative (1 of 19)
   
   → Resolution status is the strongest predictor of sentiment.

   4. Delivery is the most common complaint category.
   - Delivery issues: 15 of 30 (50% of all feedback)
   - Followed by Customer Service: 6
   - Product Quality: 6
   - Payment: 2
   
   → Half of all customer feedback mentions delivery problems.

   5. Nigeria has the highest volume of negative feedback.
   - Nigeria: 6 negative out of 11 records (54.5% negative rate)
   - Ghana: 4 negative out of 7 records (57.1% negative rate)
   - Kenya: 1 negative out of 7 records (14.3% negative rate)
   - South Africa: 1 negative out of 5 records (20.0% negative rate)
   
   → Ghana has the highest negative RATE, Nigeria has the highest VOLUME.
   → Kenya and South Africa show significantly better sentiment.

   6. Social Media is the channel with the highest proportion of negative feedback.
   - Social Media: 6 negative of 6 total (100% negative)
   - Website: 3 negative of 11 total (27.3% negative)
   - App: 3 negative of 7 total (42.9% negative)
   - WhatsApp: 0 negative of 6 total (0% negative)
   
   → Social Media complaints are overwhelmingly negative.
   → WhatsApp is the most positive channel.

   7. Low ratings strongly correlate with negative sentiment.
   - Rating 1: 5 records → all negative sentiment
   - Rating 2: 6 records → mostly negative sentiment
   - Rating 3: 4 records → all neutral sentiment
   - Rating 4: 6 records → mostly positive sentiment
   - Rating 5: 9 records → all positive sentiment
   
   → Rating is a reliable proxy for sentiment in this dataset.

   8. Faster resolution is associated with more positive sentiment.
   - Resolved in 1 day: 12 records → 11 positive, 1 negative
   - Resolved in 2-3 days: 7 records → 4 positive, 3 neutral
   - Resolved in 4+ days: 0 records (no complaints took this long)
   
   → The company resolves complaints quickly when they do resolve them.

   9. Payment failures generate strong negative sentiment.
   - Payment issues: 2 records
   - Both resulted in negative sentiment
   - FB007: "My payment failed and support did not help" (Rating 1)
   - FB020: "Payment failed and I am very disappointed" (Rating 1)
   
   → Payment failures are low in volume but high in impact.

   10. Customers with BOTH unresolved complaints AND late delivery
    are the highest-risk group.
    
    Records meeting both criteria:
    - FB002: Ghana, App, 2/5 rating, 7 days (expected 4), Unresolved
    - FB004: Nigeria, Social Media, 1/5 rating, 9 days (expected 4), Unresolved
    - FB007: Ghana, Website, 1/5 rating, 8 days (expected 4), Unresolved
    - FB011: Nigeria, Website, 2/5 rating, 7 days (expected 4), Unresolved
    - FB014: Nigeria, Social Media, 1/5 rating, 10 days (expected 4), Unresolved
    - FB020: Nigeria, App, 1/5 rating, 8 days (expected 4), Unresolved
    - FB023: Ghana, Social Media, 2/5 rating, 7 days (expected 4), Unresolved
    - FB026: Nigeria, Social Media, 1/5 rating, 9 days (expected 4), Unresolved
    - FB029: Nigeria, Website, 2/5 rating, 6 days (expected 4), Unresolved
    
    → 9 of 11 unresolved complaints also involved late delivery.
    → This combination is the single biggest driver of negative sentiment.

    11. Product category does not show strong sentiment variation.
    - Electronics: 11 records, mixed sentiment
    - Fashion: 10 records, mostly negative when delayed
    - Groceries: 9 records, mostly positive sentiment
    
    → Product category alone is not a strong predictor of sentiment.
    → Delivery experience overrides product category effect.

    12. Website and App are the primary feedback channels.
    - Website: 11 records (36.7%)
    - App: 7 records (23.3%)
    - WhatsApp: 6 records (20.0%)
    - Social Media: 6 records (20.0%)
    
    → Website is the dominant feedback channel.
    → But Social Media produces the most negative sentiment.
    
    ==================================================
    """)


# ============================================================
# RUN THE PROGRAM
# ============================================================
analyse_feedback("customer_feedback.csv")
```

*16. CUSTOMER EXPERIENCE RISK SCORE*

```python
import csv

# ==========================================================
# Function: Open file and return reader
# ==========================================================
def open_file(file):
    """ Opens a CSV file and returns a DictReader object """
    csv_file = open(file, newline="", encoding="utf-8")
    reader = csv.DictReader(csv_file)
    return reader


# ==========================================================
# RISK SCORING SETUP
# ==========================================================
rating_1 = 25
rating_2_3 = 15


delay_severe   = 20   # more than 5 days late
delay_moderate = 10   # 3 to 5 days late

unresolved_aging = 20  # unresolved and open more than 7 days
unresolved_new   = 10  # unresolved and open 7 days or less

max_score = 100


# ==========================================================
# TASK: CALCULATE RISK SCORE FOR EACH CUSTOMER
# ==========================================================
def calculate_risk(filename):
    """ Calculate a risk score for every customer feedback record """

    # --------------------------------------------------
    # Counters for the summary
    # --------------------------------------------------
    critical_count = 0
    high_count     = 0
    medium_count   = 0
    low_count      = 0

    total_score   = 0
    total_records = 0

    risk_records = []

    # --------------------------------------------------
    # Read and process each row
    # --------------------------------------------------
    reader = open_file(filename)

    for row in reader:

        score = 0

        # ----- Rating component -----
        rating = int(row["Rating"])
        if rating == 1:
            score += rating_1
        elif rating == 2 or rating == 3:
            score += rating_2_3

        # ----- Delay component -----
        actual_delivery   = int(row["Delivery_Days"])
        expected_delivery = int(row["Expected_Delivery_Days"])
        delay             = actual_delivery - expected_delivery

        if delay > 5:
            score += delay_severe
        elif delay >= 3:
            score += delay_moderate

        # ----- Resolution component -----
        resolution_status = row["Resolution_Status"]
        resolution_days   = int(row["Resolution_Days"])

        if resolution_status == "Unresolved":
            if resolution_days > 7:
                score += unresolved_aging
            else:
                score += unresolved_new

        # ----- Cap the score at the maximum -----
        if score > max_score:
            score = max_score

        # ----- Determine risk level -----
        if score >= 60:
            risk_level = "CRITICAL"
            critical_count += 1
        elif score >= 40:
            risk_level = "HIGH"
            high_count += 1
        elif score >= 20:
            risk_level = "MEDIUM"
            medium_count += 1
        else:
            risk_level = "LOW"
            low_count += 1

        # ----- Store the record for sorting -----
        risk_records.append({
            "Feedback_ID": row["Feedback_ID"],
            "Country":     row["Country"],
            "Rating":      rating,
            "Delay":       delay,
            "Status":      resolution_status,
            "Score":       score,
            "Risk_Level":  risk_level
        })

        total_score   += score
        total_records += 1

    # --------------------------------------------------
    # Sort records by score (highest first)
    # --------------------------------------------------
    risk_records.sort(key=lambda record: record["Score"], reverse=True)

    # --------------------------------------------------
    # Print header
    # --------------------------------------------------
    print("=" * 74)
    print("CUSTOMER RISK SCORE ANALYSIS")
    print("=" * 74)
    print()

    print(f"{'ID':<8}{'Country':<15}{'Rating':<8}{'Delay':<8}"
          f"{'Status':<13}{'Score':<8}{'Risk Level'}")
    print("-" * 74)

    # --------------------------------------------------
    # Print each customer record
    # --------------------------------------------------
    for record in risk_records:
        print(f"{record['Feedback_ID']:<8}"
              f"{record['Country']:<15}"
              f"{record['Rating']:<8}"
              f"{record['Delay']:<8}"
              f"{record['Status']:<13}"
              f"{record['Score']:<8}"
              f"{record['Risk_Level']}")

    # --------------------------------------------------
    # Calculate and print summary
    # --------------------------------------------------
    average_score = total_score / total_records if total_records > 0 else 0

    print("-" * 74)
    print(f"Total Customers Analysed : {total_records}")
    print(f"Average Risk Score       : {average_score:.2f} / {max_score}")
    print(f"CRITICAL (60 and above)  : {critical_count}")
    print(f"HIGH     (40 - 59)       : {high_count}")
    print(f"MEDIUM   (20 - 39)       : {medium_count}")
    print(f"LOW      (0 - 19)        : {low_count}")
    print("=" * 74)


# ==========================================================
# RUN THE PROGRAM
# ==========================================================
calculate_risk("customer_feedback.csv")
```
